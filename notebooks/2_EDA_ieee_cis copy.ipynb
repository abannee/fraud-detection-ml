```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

pd.set_option('display.max_columns', 100)
pd.set_option('display.float_format', '{:.4f}'.format)
```

```python
# 1. Load Data (Using a sample for EDA efficiency)
BASE_DIR = '/Users/abannee/Documents/GitHub/fraud_detection_ml/'
DATA_DIR = os.path.join(BASE_DIR, 'data/raw')
df = pd.read_csv(os.path.join(DATA_DIR, 'train_transaction.csv'))
```

### Data Profiling

```python
# Calculate the percentage of missing values for each column
missing_values = df.isnull().mean() * 100
# Calculate how many unique values exist per column
unique_counts = df.nunique()

# Create a summary dataframe to view our findings
quality_report = pd.DataFrame({
    'Missing_Pct': missing_values,
    'Unique_Values': unique_counts
})

# Flag columns for further inspection
missing_cols = quality_report[quality_report['Missing_Pct'] > 80].index.tolist()
```

### Feature Engineering Loop

```python
# 1. Create a dictionary to hold the new columns
new_features = {}

for col in missing_cols:
    threshold = df[df['isFraud'] == 0][col].quantile(0.99)
    flag_name = f'{col}_extreme'
    
    # Store the new series in a dictionary instead of assigning to df directly
    new_features[flag_name] = (df[col] > threshold).astype(int)

# 2. Convert dictionary to a DataFrame
new_features_df = pd.DataFrame(new_features)

# 3. Concatenate once
df = pd.concat([df, new_features_df], axis=1)
```
