```python
# Calculate the 95th percentile for normal transactions
threshold = df[df['isFraud'] == 0]['V153'].quantile(0.99)
print(f"99% of normal transactions are below: {threshold}")
```

```python
# Create the flag based on the 99th percentile
df['V153_extreme'] = (df['V153'] > threshold).astype(int)

# Check how many of our flagged transactions are actually fraud
fraud_capture = df[df['V153_extreme'] == 1]['isFraud'].value_counts(normalize=True)
print(fraud_capture)
```

```python
top_20_missing_cols = quality_report.sort_values(by='Missing_Pct', ascending=False).head(20)
print(top_20_missing_cols.index)
```

```python
missing_cols = top_20_missing_cols.index.tolist()
print(missing_cols)
```

```python
for col in missing_cols:
    threshold = df[df['isFraud'] == 0][col].quantile(0.99)
    print(f"for {col} 99% of normal transactions are below: {threshold}")

    # Create the flag based on the 99th percentile
    df[f'{col}_extreme'] = (df[col] > threshold).astype(int)


    fraud_capture = df[df[f'{col}_extreme'] == 1]['isFraud'].value_counts(normalize=True)
    print(fraud_capture)
```

```python
# Create a dictionary to store our results
fraud_impact = {}

for col in missing_cols:
    # Use your quantile logic
    threshold = df[df['isFraud'] == 0][col].quantile(0.99)
    flag_name = f'{col}_extreme'
    df[flag_name] = (df[col] > threshold).astype(int)
    
    # Capture only the fraud proportion (where isFraud == 1)
    # We use .get(1, 0) to handle cases where no fraud is captured
    impact = df[df[flag_name] == 1]['isFraud'].value_counts(normalize=True).get(1, 0)
    fraud_impact[col] = impact

# Sort and print the most impactful ones
sorted_impact = sorted(fraud_impact.items(), key=lambda x: x[1], reverse=True)
print(sorted_impact)
```
